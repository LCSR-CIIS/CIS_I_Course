```python
import sys
sys.path.append('..')  # cis1/ root

from src import Tracker, Frame, vct3, Rot
import numpy as np

tracker = Tracker("example_tracker")
frame = Frame(Rot(matrix=np.eye(3)), vct3(0, 0, 0))
cov = np.eye(3) 
tracker.create(frame, cov)
print(tracker)
```

```python
# set pose method
frame = Frame(Rot(axis='x', angle=30), vct3(1, 1, -1))
tracker.set_pose(frame)
```

```python
# set covariance method
cov = [[1, 0, 0], [0, -1, 0], [0, 0, -1]]
tracker.set_cov(cov)
```

```python
# add markers from the marker body object
markers = [vct3(1, 1, 1), vct3(2, 2, 2), vct3(3, 3, 3)]
tracker.read_markers(markers)
```
